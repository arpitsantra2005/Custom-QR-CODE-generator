```python
import qrcode
from IPython.display import display

data = "https://chatgpt.com/"

qr = qrcode.QRCode(
    version=1,
    box_size=10,
    border=4
)

qr.add_data(data)
qr.make(fit=True)

img = qr.make_image(fill_color="white", back_color="red")

display(img)

img.save("colored_qr.png")

print("Colored QR Code generated and saved as colored_qr.png")
```
